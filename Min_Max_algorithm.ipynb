Initialize and display the board

```python
def display_board(board):
    """Display the current state of the board in a 3x3 grid."""
    print()  # blank line for spacing
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()  # blank line for spacing

# Example usage:
empty_board = [' '] * 9
display_board(empty_board)
```

Function to check for a win or a draw

```python
def check_winner(board):
    """Return 'X' or 'O' if there is a winner, or None otherwise."""
    win_combos = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
        (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
        (0, 4, 8), (2, 4, 6)              # diagonals
    ]
    for (i, j, k) in win_combos:
        if board[i] == board[j] == board[k] and board[i] != ' ':
            return board[i]
    return None

def is_draw(board):
    """Return True if the board is full (no empty spaces) with no winner."""
    return ' ' not in board
```

Minimax algorithm implementation

```python
HUMAN = 'X'
AI = 'O'

def minimax(board, player):
    """Recursively compute the minimax score for the board given the current player."""
    winner = check_winner(board)
    if winner == AI:
        return 1
    elif winner == HUMAN:
        return -1
    elif is_draw(board):
        return 0

    if player == AI:
        best_score = -float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = AI
                score = minimax(board, HUMAN)
                board[i] = ' '
                best_score = max(best_score, score)
        return best_score
    else:
        best_score = float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = HUMAN
                score = minimax(board, AI)
                board[i] = ' '
                best_score = min(best_score, score)
        return best_score

def get_best_move(board):
    """Find the best move for AI by running minimax on each possible move."""
    best_score = -float('inf')
    best_move = None
    for i in range(9):
        if board[i] == ' ':
            board[i] = AI
            score = minimax(board, HUMAN)
            board[i] = ' '
            if score > best_score:
                best_score = score
                best_move = i
    return best_move
```

 Function to get a valid human move

```python
def get_human_move(board):
    """Prompt the human player to enter a move and return the valid index (0-8)."""
    while True:
        try:
            pos = int(input("Enter your move (1-9): "))
            if pos < 1 or pos > 9:
                print("Invalid input: Choose a number from 1 to 9.")
                continue
            idx = pos - 1
            if board[idx] != ' ':
                print("That cell is already taken. Choose another move.")
                continue
            return idx
        except ValueError:
            print("Invalid input: Please enter a number.")
```

Main game loop with scoreboard

```python
def play_game():
    board = [' '] * 9
    current_player = HUMAN  # Human plays 'X' and goes first
    while True:
        display_board(board)

        if current_player == HUMAN:
            idx = get_human_move(board)
            board[idx] = HUMAN
        else:
            idx = get_best_move(board)
            board[idx] = AI
            print(f"AI chooses position {idx+1}")

        # Check for end of game
        winner = check_winner(board)
        if winner or is_draw(board):
            display_board(board)
            if winner == HUMAN:
                print("Congratulations! You win!")
                return "human"
            elif winner == AI:
                print("AI wins. Better luck next time!")
                return "ai"
            else:
                print("It's a draw!")
                return "draw"

        # Switch player
        current_player = AI if current_player == HUMAN else HUMAN

# Initialize scoreboard
scores = {"human": 0, "ai": 0, "draw": 0}

while True:
    result = play_game()
    # Update scoreboard
    if result == "human":
        scores["human"] += 1
    elif result == "ai":
        scores["ai"] += 1
    else:
        scores["draw"] += 1

    # Print running scoreboard
    print(f"Scoreboard: You {scores['human']} - AI {scores['ai']} - Draws {scores['draw']}")

    # Ask to play again
    again = input("Play again? (Y/N): ").strip().lower()
    if again != 'y':
        print("Thank you for playing!")
        break
```
